```python
import tensorflow as tf;
from tensorflow.keras.models import load_model;
import pickle
import pandas as pd;
import numpy as np;
```

```python
model=load_model('model.h5');
with open('label_encoder_gender.pkl','rb') as file:
    oneHot_encoder_gen=pickle.load(file)
```

```python
with open('label_encoder_geography.pkl','rb') as file:
    label_encoder_geo=pickle.load(file)
```

```python
with open('scalar.pkl','rb') as file:
    scalar=pickle.load(file)
```

```python
input_data={
    'CreditScore': 600,
    'Geography' : 'France',
    'Gender' : 'Male',
    'Age' : 40,
    'Tenure' : 3,
    'Balance' : 60000,
    'NumOfProducts' : 2,
    'HasCrCard' : 1,
    'IsActiveMember' : 1,
    'EstimatedSalary' : 50000
}
```

```python
input_data['Geography']
```

```python
geo_encoded=label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=label_encoder_geo.get_feature_names_out(['Geography']))
```

```python
geo_encoded_df
```

```python
input_data=pd.DataFrame([input_data])
```

```python
input_data
```

```python
input_data['Gender']=oneHot_encoder_gen.transform(input_data['Gender'])
```

```python
input_data
```

```python
input_data_df=pd.concat([input_data,geo_encoded_df],axis=1)
```

```python
input_data_df.drop(columns=['Geography'],inplace=True)
```

```python
input_data_df
```

```python
input_data_df_scaled=scalar.transform(input_data_df)
```

```python
input_data_df_scaled
```

```python
predict=model.predict(input_data_df_scaled)
```
